```python
from numba import cuda
numba.cuda.detect()

device = numba.cuda.select_device(0)

_, total_mem = cuda.current_context().get_memory_info()

core_count = device.MULTIPROCESSOR_COUNT * 64 

print("Device ID:", device.id)
print("Device Name:", device.name)
print("Number of Processor:", device.MULTIPROCESSOR_COUNT)
print("Device's cores:", core_count)
print(f"Memory Size: {total_mem / 1024**2:.2f} MB")
```
